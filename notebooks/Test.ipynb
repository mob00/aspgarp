# Simulate a model

Choose `tree`, `vessels`, `population`, `falling`, or `heating`. The result is the full observable state graph, including self-loops. Simulation requires the `clingo` Python package.

````python
import sys
from pathlib import Path

working_directory = Path.cwd()
for root in (working_directory, *working_directory.parents):
    if (root / "src/aspgarp/simulation.py").is_file():
        break
else:
    raise RuntimeError("Open this notebook from the ASPGARP checkout.")

sys.path.insert(0, str(root / "src"))
from aspgarp.simulation import simulate
from aspgarp.export import to_facts
````

````python
model_folder = root / "models" / "tree"
graph = simulate(model_folder)
print(f"{len(graph.states)} states, {len(graph.transitions)} transitions")
print(to_facts(graph), end="")
````

## Run from the shell

From the repository root, run either input form.

```bash
python main.py models/tree
python main.py models/tree/db.lp models/tree/ref.lp --format dot -o tree.dot
dot -Tsvg tree.dot -o tree.svg
```

DOT export requires Clingraph. To generate the same DOT directly in Python, import `to_dot` from `aspgarp.export` and call `to_dot(graph)`.
